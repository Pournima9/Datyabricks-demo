##Setup the project environment for formula1 project
##### Create external location - databrickscourseextstg

###### Access Cloud Storage

```sql
%fs ls 'abfss://formula1@databrickscourseextstg.dfs.core.windows.net/landing'
```

###### Create external location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databrickscourseextstg_formula1
URL 'abfss://formula1@databrickscourseextstg.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-course-stg-cred`)
COMMENT 'External location for the formula1 container';
```

######Create Catalogs

```sql
SHOW CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS formula1
MANAGED LOCATION 'abfss://formula1@databrickscourseextstg.dfs.core.windows.net/'
COMMENT 'This is the main catalog for formula1 project'
```

###### Create schemas, landing, bronze, silver, gold

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;

CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@databrickscourseextstg.dfs.core.windows.net/bronze';

CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@databrickscourseextstg.dfs.core.windows.net/silver';

CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@databrickscourseextstg.dfs.core.windows.net/gold';

    
```

```sql
SHOW SCHEMAS;

USE CATALOG formula1;
```

###### Create Volume


```sql
CREATE EXTERNAL VOLUME formula1.landing.files
LOCATION 'abfss://formula1@databrickscourseextstg.dfs.core.windows.net/landing';
```
